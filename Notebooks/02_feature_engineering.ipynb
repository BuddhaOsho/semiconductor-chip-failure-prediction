```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import sklearn

from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, OrdinalEncoder

```

```python
df = pd.read_csv("../data/semiconductor_chip_data.csv")
df.head()
df= df.drop(columns=["chip_id","risk_level"])
```

```python
X = df.drop(columns=["chip_failure"])
Y=df["chip_failure"]
```

#Separating the data as per the categories

```python
Numerical_Data=["temperature_c","voltage_v","frequency_mhz","current_a","power_consumption_mw","leakage_current_ua","test_duration_sec","defect_rate"]

Ordinal_Data=  ["defect_severity"]

Nominal_Data = ["manufacturing_process"]
```

#Creating Diffrent Pipeline Transformer

#Combining Pipeline Transformer with thier respective data


```python
from sklearn.compose import ColumnTransformer
```

```python
Numerical_Pipeline = Pipeline(steps=[
    ("impute", SimpleImputer(strategy="mean")),
    ("scaling", StandardScaler())
])

Ordinal_Pipeline = Pipeline(steps=[
    ("impute", SimpleImputer(strategy="most_frequent")),
    ("encoding", OrdinalEncoder(categories=[["Low", "Medium", "High",'Critical']])),
    ("scaling", StandardScaler())
])

Nominal_Pipeline = Pipeline(steps=[
    ("impute", SimpleImputer(strategy="most_frequent")),
    ("encoding", OneHotEncoder(handle_unknown="ignore"))
])
```

```python
preprocessor = ColumnTransformer(transformers=[
    ("Numerical_Transformation", Numerical_Pipeline, Numerical_Data),
    ("Ordinal_Transformation", Ordinal_Pipeline, Ordinal_Data),
    ("Nominal_Transformation", Nominal_Pipeline, Nominal_Data)
])
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
```

```python
le = LabelEncoder()
```

```python
Y = le.fit_transform(Y)
```

```python
X_train,X_test, Y_train, Y_test = train_test_split(X,Y, test_size=0.2,  random_state=42)
```

```python
import joblib

joblib.dump(preprocessor, "../models//preprocessor.pkl")

joblib.dump(X_train, "../artifacts/X_train.pkl")
joblib.dump(X_test, "../artifacts/X_test.pkl")

joblib.dump(Y_train, "../artifacts/y_train.pkl")
joblib.dump(Y_test, "../artifacts/y_test.pkl")

joblib.dump(preprocessor, "../artifacts/preprocessor.pkl")
```
